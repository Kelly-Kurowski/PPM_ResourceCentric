```python
import os
import json
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# --------------------------
# User configuration
# --------------------------
# Map encodings to their respective folders containing JSON results
encoding_folders = {
    "Encoding 1": "BPIC18/LSTM model/Baseline encoding",
    "Encoding 2": "BPIC18/LSTM model/SCap_check",
    "Encoding 3": "BPIC18/LSTM model/S2g",
    "Encoding 4": "BPIC18/LSTM model/S2gR"
}

# --------------------------
# Read JSON results
# --------------------------
def read_accuracy_from_folder(folder_path):
    accuracies = []
    prefixes = []
    # Loop over all JSON files in the folder
    for file_name in sorted(os.listdir(folder_path)):
        if file_name.endswith(".json"):
            full_path = os.path.join(folder_path, file_name)
            with open(full_path, "r") as f:
                data = json.load(f)
                # Extract sequence length and accuracy
                prefixes.append(data["sequence_length"])
                accuracies.append(data["metrics"]["accuracy"])
    return prefixes, accuracies

# --------------------------
# Collect all results
# --------------------------
all_data = []
for encoding, folder in encoding_folders.items():
    prefixes, accuracies = read_accuracy_from_folder(folder)
    for p, a in zip(prefixes, accuracies):
        all_data.append({
            "Prefix Length": p,
            "Accuracy": a,
            "Encoding": encoding
        })

# Create DataFrame
df_plot = pd.DataFrame(all_data)

# --------------------------
# Plot
# --------------------------
plt.figure(figsize=(12, 8))
sns.lineplot(
    data=df_plot, 
    x="Prefix Length", 
    y="Accuracy", 
    hue="Encoding", 
    marker="o", 
    linewidth=2
)

plt.title("LSTM", fontsize=24)
plt.ylim(0, 1)
plt.grid(True)

# X-axis formatting
unique_prefixes = sorted(df_plot["Prefix Length"].unique())
plt.xticks(ticks=unique_prefixes, rotation=45, fontsize=20)
plt.yticks(fontsize=20)
plt.xlabel("Prefix Length", fontsize=22)
plt.ylabel("Accuracy", fontsize=22, labelpad=10)

# Legend below the plot
plt.legend(loc="upper center", bbox_to_anchor=(0.5, -0.2), ncol=len(encoding_folders), fontsize=12)
plt.subplots_adjust(bottom=0.15)
plt.tight_layout()

# --------------------------
# Save plot
# --------------------------
download_path = os.path.expanduser('~/Downloads/accuracy_plot_LSTM_18.svg')
plt.savefig(download_path, format='svg')
print(f"✅ Plot saved to: {download_path}")

# Show plot
plt.show()
```

```python
import os
import json
import numpy as np
import pandas as pd

# ---------------------------------------
# Configuration
# ---------------------------------------
results_root = "."  # or use "results" if your logs are in ./results/...
datasets = ["BPIC13", "BPIC17", "BPIC18", "BPIC19"]

# Use your exact folder names here:
models = ["LightGBM", "LSTM model", "RandomForest", "Transformer model"]

# Encodings (folder names inside each model directory)
encodings = ["Baseline encoding", "SCap", "S2g", "S2gR"]

# ---------------------------------------
# Helper: Read accuracies from JSONs
# ---------------------------------------
def read_accuracies(folder_path):
    accuracies = []
    if not os.path.exists(folder_path):
        return []
    for file_name in os.listdir(folder_path):
        if file_name.endswith(".json"):
            full_path = os.path.join(folder_path, file_name)
            with open(full_path, "r") as f:
                try:
                    data = json.load(f)
                    acc = data.get("metrics", {}).get("accuracy", None)
                    if acc is not None:
                        accuracies.append(acc)
                except Exception as e:
                    print(f"⚠️ Error reading {file_name}: {e}")
    return accuracies

# ---------------------------------------
# Collect results
# ---------------------------------------
rows = []
for dataset in datasets:
    for model in models:
        results = {}
        for enc in encodings:
            folder = os.path.join(results_root, dataset, model, enc)
            accs = read_accuracies(folder)
            if accs:
                mean = np.mean(accs)
                std = np.std(accs)
                results[enc] = f"{mean:.2f} ± {std:.2f}"
            else:
                results[enc] = "--"
        rows.append({
            "Dataset": dataset,
            "Model": model,
            "Baseline encoding": results["Baseline encoding"],
            "SCap": results["SCap"],
            "S2g": results["S2g"],
            "S2gR": results["S2gR"]
        })

# ---------------------------------------
# Create DataFrame and display
# ---------------------------------------
df = pd.DataFrame(rows)

print("\n✅ Summary Table (copy this directly into Word):\n")
print(df.to_string(index=False))

# ---------------------------------------
# Save table as CSV
# ---------------------------------------
output_path = "accuracy_summary_table.csv"
df.to_csv(output_path, index=False)
print(f"\n💾 Saved table to: {output_path}")
```
